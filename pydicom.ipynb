```python
import os
import pydicom

def modify_dicom_tags(directory, new_study_instance_uid, new_patient_id, new_accession_number):
    # Recorre todos los archivos en el directorio
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith(".dcm"):
                file_path = os.path.join(root, file)
                try:
                    # Lee el archivo DICOM
                    ds = pydicom.dcmread(file_path)
                    
                    # Modifica el Study Instance UID
                    ds.StudyInstanceUID = new_study_instance_uid
                    
                    # Modifica el Patient ID
                    ds.PatientID = new_patient_id
                    
                    # Modifica el Accession Number
                    ds.AccessionNumber = new_accession_number
                    
                    # Guarda el archivo DICOM modificado
                    ds.save_as(file_path)
                    print(f"Modificado: {file_path}")
                except Exception as e:
                    print(f"Error al modificar {file_path}: {e}")

# Configura el directorio y los nuevos valores de los tags DICOM
directory = "D:\Softinhealth\ImplementacionPACS\IMAGENES\DBT01"
new_study_instance_uid = "1.2.826.0.1.3680043.8.498.59396554999264091294352192820758901394"
new_patient_id = "NR52111"
new_accession_number = "ACC003"

# Modifica los tags DICOM en todos los archivos del directorio
modify_dicom_tags(directory, new_study_instance_uid, new_patient_id, new_accession_number)
```
