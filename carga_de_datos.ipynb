```python
import os
import shutil

# Directorio de origen (donde se encuentran los archivos)
directorio_origen = "D:\MEDICAL IT\CLIENTES\SDLM\W3SVC1"

# Directorio de destino (donde se copiarán los archivos)
directorio_destino = "D:\MEDICAL IT\CLIENTES\SDLM\LogsDiasClaves\8horas"

# Lista de cadenas que deben estar al final del nombre del archivo
filtros = ["07.log", "08.log", "09.log", "10.log"]

# Recorrer el directorio de origen
for archivo in os.listdir(directorio_origen):
    # Verificar si el nombre del archivo termina con alguna de las cadenas en la lista
    if any(archivo.endswith(filtro) for filtro in filtros):
        # Copiar el archivo al directorio de destino
        shutil.copy(os.path.join(directorio_origen, archivo), directorio_destino)

        # Imprimir un mensaje informativo
        print(f"Archivo copiado: {archivo}")
```

```python
import openpyxl

# Ruta del archivo Excel
archivo_xlsx = 'predef.xlsx'

# Abrir el archivo de Excel
workbook = openpyxl.load_workbook(archivo_xlsx)

# Lista para almacenar las consultas
querys = []

# Iterar sobre todas las hojas del archivo
for sheet in workbook.sheetnames:
    # Verificar si el nombre de la hoja comienza con "Patrones"
    if sheet.startswith("PatronesUS"):
        worksheet = workbook[sheet]
        print(f"\nHoja: {sheet}")
        
        # Iterar sobre las filas de la hoja
        for row in worksheet.iter_rows(min_row=2, max_row=worksheet.max_row):
            query = f"""INSERT INTO public.tbreportdefault (guid, code, name, idmodality, executiontext, findingtext, conclusiontext) 
                        VALUES (uuid_generate_v4(), '{row[0].value}', '{row[2].value}', 
                                (SELECT ARRAY[Guid] FROM public.ismodality WHERE Description = '{row[1].value}'), 
                                '{row[3].value}', '{row[4].value}', '{row[5].value}');"""
            querys.append(query)

# Guardar las consultas en un archivo de texto con codificación utf-8
with open('querys.txt', 'w', encoding='utf-8') as file:
    for query in querys:
        file.write(query + '\n')

print("Consultas guardadas en querys.txt")


```
